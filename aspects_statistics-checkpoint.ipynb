```python
# !pip install transformers torch rake-nltk pandas matplotlib -q

from transformers import pipeline
import pandas as pd
from rake_nltk import Rake
import matplotlib.pyplot as plt
```

```python
# ------------------------------
# 1. Пример отзывов (можно заменить на реальные)
# ------------------------------
reviews = [
    "Доктор очень внимательный и всё объяснил, но ждать пришлось больше часа.",
    "В регистратуре хамят, а сами врачи хорошие.",
    "Приём прошёл быстро, но в коридоре было грязно.",
    "Лечение помогло, но оборудование в кабинете старое.",
    "Записаться на приём трудно, телефон постоянно занят."
]
```

```python
# ------------------------------
# 2. Модель для анализа тональности
# ------------------------------
sentiment_analyzer = pipeline("sentiment-analysis", model="blanchefort/rubert-base-cased-sentiment")
```

```python
# ------------------------------
# 3. Функция извлечения аспектов
# ------------------------------
def extract_aspects(text, top_n=2):
    r = Rake(language="russian")
    r.extract_keywords_from_text(text)
    return [kw for score, kw in r.get_ranked_phrases_with_scores()[:top_n]]
```

```python
# ------------------------------
# 4. Обработка отзывов
# ------------------------------
data = []
for review in reviews:
    sentiment = sentiment_analyzer(review)[0]
    aspects = extract_aspects(review, top_n=2)
    for aspect in aspects:
        data.append({
            "review": review,
            "aspect": aspect.strip(),
            "sentiment": sentiment["label"],
            "score": round(sentiment["score"], 3)
        })

df = pd.DataFrame(data)

print("🔎 Обработанные аспекты с тональностью:")
display(df)
```

```python
# ------------------------------
# 5. Агрегация статистики
# ------------------------------
stats = df.groupby(["aspect", "sentiment"]).size().unstack(fill_value=0)
stats["Всего отзывов"] = stats.sum(axis=1)
percent_stats = stats.div(stats["Всего отзывов"], axis=0) * 100

print("📊 Агрегированная статистика по аспектам (%):")
display(percent_stats.round(1))
```

```python
# ------------------------------
# 6. Визуализация
# ------------------------------
percent_stats[["POSITIVE","NEGATIVE"]].plot(kind="bar", stacked=True, figsize=(10,5))
plt.title("Распределение тональности по аспектам")
plt.ylabel("% отзывов")
plt.xlabel("Аспекты")
plt.legend(title="Тональность")
plt.show()
```
